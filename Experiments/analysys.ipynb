## Motor air 3508 M1

```python
#motor 1 m3508 - exp. 1
data_air_m3508_m1_1 = pd.read_csv('air/AIR - 3508 - M1 - 1- 12120918.csv')
data_air_m3508_m1_2 = pd.read_csv('air/AIR - 3508 - M1 - 2 - 12120939.csv')
data_air_m3508_m2_1 = pd.read_csv('air/AIR - 3508 - M2 -1 - 12121610.csv')
data_air_m3508_m2_2 = pd.read_csv('air/AIR - 3508 - M2 - 2 - 12121642.csv')
df_air_m3508_m1_1 = pd.DataFrame(data_air_m3508_m1_1)
df_air_m3508_m1_2 = pd.DataFrame(data_air_m3508_m1_2)
df_air_m3508_m2_1 = pd.DataFrame(data_air_m3508_m2_1)
df_air_m3508_m2_2 = pd.DataFrame(data_air_m3508_m2_2)
```

```python
#Add a column for mechanical power, where Mechanical_Power = torque * 2 * rpm * pi / 60
# Convert comma decimal separator to dot and then to float for numeric columns
cols_to_convert = ['Torque(Nm)', 'RPM', 'AC1(A)', 'AC2(A)', 'AC3(A)', 'Electrical power(W)', 'Thrust(Kg)']
for col in cols_to_convert:
  df_air_m3508_m1_1[col] = df_air_m3508_m1_1[col].astype(str).str.replace(',', '.').astype(float)
  df_air_m3508_m1_2[col] = df_air_m3508_m1_2[col].astype(str).str.replace(',', '.').astype(float)
  df_air_m3508_m2_1[col] = df_air_m3508_m2_1[col].astype(str).str.replace(',', '.').astype(float)
  df_air_m3508_m2_2[col] = df_air_m3508_m2_2[col].astype(str).str.replace(',', '.').astype(float)

# Add Mechanical Power column: P = Torque * 2 * pi * RPM / 60
df_air_m3508_m1_1['Mechanical Power(W)'] = df_air_m3508_m1_1['Torque(Nm)'] * 2 * np.pi * df_air_m3508_m1_1['RPM'] / 60
df_air_m3508_m1_2['Mechanical Power(W)'] = df_air_m3508_m1_2['Torque(Nm)'] * 2 * np.pi * df_air_m3508_m1_2['RPM'] / 60
df_air_m3508_m2_1['Mechanical Power(W)'] = df_air_m3508_m2_1['Torque(Nm)'] * 2 * np.pi * df_air_m3508_m2_1['RPM'] / 60
df_air_m3508_m2_2['Mechanical Power(W)'] = df_air_m3508_m2_2['Torque(Nm)'] * 2 * np.pi * df_air_m3508_m2_2['RPM'] / 60
```

```python
print(df_air_m3508_m1_1.columns)
```

```python
#mean
# Motor Model | Test Number |Thrust(Kg) |Max Torque (Nm) | Max RPM |AC1(A) | AC2(A) | AC3(A) | Max Mechanical Power (W) | Max Electrical Power (W) | Efficiency at Max Power (%) 
#PARA THROTTLE DE 20 40 60 80%
throttle_levels = [20, 40, 60, 80]
results = []
for throttle in throttle_levels:
    for df, motor_model, test_number in [
        (df_air_m3508_m1_1, 'M3508', 'M1 Test 1'),
        (df_air_m3508_m1_2, 'M3508', 'M1 Test 2'),
        (df_air_m3508_m2_1, 'M3508', 'M2 Test 1'),
        (df_air_m3508_m2_2, 'M3508', 'M2 Test 2')
    ]:
        # Corrected column name to 'Throttle(%)'
        df_throttle = df[df['Throttle(%)'] == throttle]
        if not df_throttle.empty:
            mean_thrust = df_throttle['Thrust(Kg)'].mean()
            mean_torque = df_throttle['Torque(Nm)'].mean()
            mean_rpm = df_throttle['RPM'].mean()
            mean_ac1 = df_throttle['AC1(A)'].mean()
            mean_ac2 = df_throttle['AC2(A)'].mean()
            mean_ac3 = df_throttle['AC3(A)'].mean()
            mean_mech_power = df_throttle['Mechanical Power(W)'].mean()
            mean_elec_power = df_throttle['Electrical power(W)'].mean()
            efficiency = (mean_mech_power / mean_elec_power * 100) if mean_elec_power > 0 else 0

            results.append({
                'Motor Model': motor_model,
                'Test Number': test_number,
                'Throttle (%)': throttle,
                'Thrust (Kg)': mean_thrust,
                'Max Torque (Nm)': mean_torque,
                'Max RPM': mean_rpm,
                'AC1 (A)': mean_ac1,
                'AC2 (A)': mean_ac2,
                'AC3 (A)': mean_ac3,
                'Max Mechanical Power (W)': mean_mech_power,
                'Max Electrical Power (W)': mean_elec_power,
                'Efficiency at Max Power (%)': efficiency
            })
summary_df = pd.DataFrame(results)
summary_df
```

```python
# Increase global font size
plt.rcParams.update({'font.size': 28})

# Datasets to plot
datasets = [
    (df_air_m3508_m1_1, 'M1 Test 1'),
    (df_air_m3508_m1_2, 'M1 Test 2'),
    (df_air_m3508_m2_1, 'M2 Test 1'),
    (df_air_m3508_m2_2, 'M2 Test 2')
]

# Create subplots
fig, axs = plt.subplots(3, 2, figsize=(25, 25), dpi=300)
#fig.suptitle('M3508 Analysis: Variables vs Throttle (0-80%)', fontsize=36)

# Iterate through datasets and plot
for df, label in datasets:
    # Filter for throttle <= 80% and sort
    # Consider only the ascent (up to throttle peak)
    max_idx = df['Throttle(%)'].idxmax()
    df_ascending = df.loc[:max_idx]
    df_filtered = df_ascending[df_ascending['Throttle(%)'] <= 80].sort_values('Throttle(%)')
    
    if not df_filtered.empty:
        x = df_filtered['Throttle(%)']
        
        # Helper to plot scatter with R2 in legend
        def plot_scatter(ax, y, title, ylabel, custom_label=None):
            if len(x) > 1:
                # Calculate R2 (Linear)
                r2 = np.corrcoef(x, y)[0, 1]**2
                lbl = f"{custom_label if custom_label else label} ($R^2={r2:.3f}$)"
            else:
                lbl = custom_label if custom_label else label
            ax.scatter(x, y, label=lbl, s=100) # Increased point size
            ax.set_title(title, fontsize=32)
            ax.set_xlabel('Throttle (%)', fontsize=28)
            ax.set_ylabel(ylabel, fontsize=28)
            ax.tick_params(axis='both', which='major', labelsize=26)
            ax.grid(True)

        # Torque
        plot_scatter(axs[0, 0], df_filtered['Torque(Nm)'], 'Torque vs Throttle', 'Torque (Nm)')
        
        # RPM
        plot_scatter(axs[0, 1], df_filtered['RPM'], 'RPM vs Throttle', 'RPM')
        
        # AC Current (Phase 1 only)
        plot_scatter(axs[1, 0], df_filtered['AC1(A)'], 'AC Current (Phase 1) vs Throttle', 'Current (A)')
        
        # Electrical Power
        plot_scatter(axs[1, 1], df_filtered['Electrical power(W)'], 'Electrical Power vs Throttle', 'Power (W)')
        
        # Thrust
        plot_scatter(axs[2, 0], df_filtered['Thrust(Kg)'], 'Thrust vs Throttle', 'Thrust (Kg)')
        
        # Efficiency
        efficiency = df_filtered['Mechanical Power(W)'] / df_filtered['Electrical power(W)'] * 100
        plot_scatter(axs[2, 1], efficiency, 'Efficiency vs Throttle', 'Efficiency (%)')

# Add legends
for ax in axs.flat:
    ax.legend(fontsize=24)

plt.tight_layout()
plt.savefig('m3508_analysis.png', dpi=300)
plt.show()

```

```python
#table
# Motor Model | Test Number |Thrust(Kg) |Max Torque (Nm) | Max RPM |AC1(A) | AC2(A) | AC3(A) | Max Mechanical Power (W) | Max Electrical Power (W) | Efficiency at Max Power (%) 
#for THROTTLE DE 20 40 60 80%

throttle_levels = [20, 40, 60, 80]
results = []
for throttle in throttle_levels:
        # Corrected column name to 'Throttle(%)'
        df_throttle = df_air_m3508_f[df_air_m3508_f['Throttle(%)'] == throttle]
        if not df_throttle.empty:
            max_thrust = df_throttle['Thrust(Kg)'].max()
            max_torque = df_throttle['Torque(Nm)'].max()
            max_rpm = df_throttle['RPM'].max()
            max_ac1 = df_throttle['AC1(A)'].max()
            max_ac2 = df_throttle['AC2(A)'].max()
            max_ac3 = df_throttle['AC3(A)'].max()
            max_mech_power = df_throttle['Mechanical Power(W)'].max()
            max_elec_power = df_throttle['Electrical power(W)'].max()
            efficiency = (max_mech_power / max_elec_power * 100) if max_elec_power > 0 else 0

            results.append({
                'Motor Model': 'M3508',
                'Test Number': 'Fail Test',
                'Throttle (%)': throttle,
                'Thrust (Kg)': max_thrust,
                'Max Torque (Nm)': max_torque,
                'Max RPM': max_rpm,
                'AC1 (A)': max_ac1,
                'AC2 (A)': max_ac2,
                'AC3 (A)': max_ac3,
                'Max Mechanical Power (W)': max_mech_power,
                'Max Electrical Power (W)': max_elec_power,
                'Efficiency at Max Power (%)': efficiency
            })
summary_df_fail = pd.DataFrame(results)
summary_df_fail
```

```python
# Increase global font size
plt.rcParams.update({'font.size': 28})

# Datasets to plot
datasets = [
    (df_air_m3508_f, 'M3508 Fail Test')
]

# Create subplots
fig, axs = plt.subplots(3, 2, figsize=(25, 25), dpi=300)
#fig.suptitle('M3508 Fail Analysis: Variables vs Throttle (0-80%)', fontsize=36)

# Iterate through datasets and plot
for df, label in datasets:
    # Filter for throttle <= 80% and sort
    # Consider only the ascent (up to throttle peak)
    max_idx = df['Throttle(%)'].idxmax()
    df_ascending = df.loc[:max_idx]
    df_filtered = df_ascending[df_ascending['Throttle(%)'] <= 80].sort_values('Throttle(%)')
    
    if not df_filtered.empty:
        x = df_filtered['Throttle(%)']
        
        # Helper to plot scatter with R2 in legend
        def plot_scatter(ax, y, title, ylabel, custom_label=None):
            if len(x) > 1:
                # Calculate R2 (Linear)
                r2 = np.corrcoef(x, y)[0, 1]**2
                lbl = f"{custom_label if custom_label else label} ($R^2={r2:.3f}$)"
            else:
                lbl = custom_label if custom_label else label
            ax.scatter(x, y, label=lbl, s=100) # Increased point size
            ax.set_title(title, fontsize=32)
            ax.set_xlabel('Throttle (%)', fontsize=28)
            ax.set_ylabel(ylabel, fontsize=28)
            ax.tick_params(axis='both', which='major', labelsize=26)
            ax.grid(True)

        # Torque
        plot_scatter(axs[0, 0], df_filtered['Torque(Nm)'], 'Torque vs Throttle', 'Torque (Nm)')
        
        # RPM
        plot_scatter(axs[0, 1], df_filtered['RPM'], 'RPM vs Throttle', 'RPM')
        
        # AC Current (Phase 1 only)
        plot_scatter(axs[1, 0], df_filtered['AC1(A)'], 'AC Current vs Throttle', 'Current (A)',custom_label='Phase 1')
        plot_scatter(axs[1, 0], df_filtered['AC2(A)'], 'AC Current vs Throttle', 'Current (A)', custom_label='Phase 2')
        plot_scatter(axs[1, 0], df_filtered['AC3(A)'], 'AC Current vs Throttle', 'Current (A)', custom_label='Phase 3')
        
        
        # Electrical Power
        plot_scatter(axs[1, 1], df_filtered['Electrical power(W)'], 'Electrical Power vs Throttle', 'Power (W)')
        
        # Thrust
        plot_scatter(axs[2, 0], df_filtered['Thrust(Kg)'], 'Thrust vs Throttle', 'Thrust (Kg)')
        
        # Efficiency
        efficiency = df_filtered['Mechanical Power(W)'] / df_filtered['Electrical power(W)'] * 100
        plot_scatter(axs[2, 1], efficiency, 'Efficiency vs Throttle', 'Efficiency (%)')

# Add legends
for ax in axs.flat:
    ax.legend(fontsize=24)

plt.tight_layout()
plt.savefig('m3508_fail_analysis.png', dpi=300)
plt.show()

```
